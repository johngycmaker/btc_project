### Actual gradio app to be run on presentation

```python
import gradio as gr
from src.inference import predict_with_latest_ohlcv
# from inference import predict_with_latest_ohlcv
```

```python
def infer_ui(sentiment):
    result = predict_with_latest_ohlcv(sentiment)

    pred_text = (
        f"Prediction: {result['prediction']}\n"
        f"Latest week used: {result['latest_week_used']}\n"
        f"Open: {result['open']}\n"
        f"Close: {result['close']}\n"
        f"High: {result['high']}\n"
        f"Low: {result['low']}\n"
        f"Volume: {result['volume']}\n"
    )

    # Optional: return a label + probability for UI clarity
    return pred_text
```

```python
demo = gr.Interface(
    fn=infer_ui,
    inputs=gr.Slider(minimum=-1, maximum=1, step=0.1, label="Weekly Sentiment", value=0.0),
    outputs=gr.Textbox(label="Result", lines=10),
    title="BTC Weekly Predictor (Sentiment-only input)",
    description="Enter sentiment. App uses latest week OHLCV from weekly_history.csv automatically."
)

demo.launch(share=True, inline=False)
```
