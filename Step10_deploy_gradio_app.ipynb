### Actual gradio app to be run on presentation

```python
import gradio as gr
from src.inference import predict_with_latest_ohlcv
```

```python
def infer_ui(sentiment):
    result = predict_with_latest_ohlcv(sentiment)

    pred_text = (
        f"================================================\n"
        f"Prediction: {result['prediction']}\n"
        f"================================================\n"
        f"\n"
        f"Latest week used: {result['latest_week_used']}\n"
        f"Open: {result['open']}\n"
        f"Close: {result['close']}\n"
        f"High: {result['high']}\n"
        f"Low: {result['low']}\n"
        f"Volume: {result['volume']}\n"
    )

    # Optional: return a label + probability for UI clarity
    return pred_text
```

```python
import gradio as gr

with gr.Blocks() as demo:
    gr.Markdown("# BTC Weekly Predictor (Sentiment-only input)")
    gr.Markdown("Enter sentiment. App uses latest week OHLCV from weekly_history.csv automatically.")

    sentiment = gr.Slider(
        minimum=-1, maximum=1, step=1,
        label="Weekly Sentiment", value=0.0
    )

    # Endpoint labels under slider
    with gr.Row():
        gr.Markdown("<span style='font-size: 0.9em;'>Negative</span>")
        gr.Markdown("<div style='text-align:right; font-size: 0.9em;'>Positive</div>")

    out = gr.Textbox(label="Result", lines=10)

    with gr.Row():
        submit_btn = gr.Button("Submit")
        clear_btn  = gr.Button("Clear")

    # Submit -> run inference
    submit_btn.click(fn=infer_ui, inputs=sentiment, outputs=out)

    # Clear -> reset slider + output
    clear_btn.click(fn=lambda: (0.0, ""), inputs=None, outputs=[sentiment, out])

demo.launch(inline=True)
```
