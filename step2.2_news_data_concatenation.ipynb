```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
# Check the current folder directory
import os
os.getcwd()
```

```python
# import only csv files for 
from pathlib import Path
folder = Path("../Project_btc_sentiment/ compile_data/")  # current folder; change to Path("some/subfolder") if needed

all_files = [p.name for p in folder.rglob("*.csv") if p.is_file()]
```

```python
# check the filename
for i, file in enumerate(all_files[:4]) : 
    print(f'file no {i+1} : {file}') 
```

```python
# Load Dataframe
dfs = [] 
for file in all_files[:4]: 
    dfs.append(pd.read_csv(f'../Project_btc_sentiment/ compile_data/{file}'))
```

```python
# check if dataframe have same schema 
def df_schema_same(df_a,df_b): 
    same_column = df_a.columns.equals(df_b.columns) 
    same_dtype = df_a.dtypes.equals(df_b.dtypes) 
    comb = same_column & same_dtype 
    return f'Both dataframe have same column and types : {comb}'
```

```python
# Check schema randomly 
for i in range (3) :
    df_a = dfs[i]
    df_b = dfs[i+1]
    result = df_schema_same(df_a, df_b)
    print(result)
    print(f'df{i+1} vs df{i+2}')
```

```python
# Begin data concatenation 
df_comb = pd.concat(dfs, axis = 0)
```

```python
# Check concatenation shape 
row_count = 0
for df in dfs : 
    row_count += len(df) 
    print(row_count)
```

## GUID (Global Unique Identifier) 
- A GUID is a unique reference number used as an identifier in computer systems. 
- The "Article guid" is unique for every article and serves as its distinct identifier 

```python
# Check the unique ID "GUID" count)
df_comb['GUID'].nunique()
```

```python
# Drop duplicate based on GUID 
df_comb_no_dup =  df_comb.drop_duplicates(subset = ['GUID'], keep = 'first')
```

```python
print(df_comb_no_dup.shape)
```

```python
min_max_date = df_comb_no_dup['date'].agg(['min','max'])
min_max_date
```

```python
# Save the concatenated file as csv
df_comb_no_dup.to_csv(f"data_{min_max_date['min']}_{min_max_date['max']}.csv")
```
